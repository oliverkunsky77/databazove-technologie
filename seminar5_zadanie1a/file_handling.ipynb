<h2 style="margin: 24px 0 16px; font-size: 20px; font-weight: 600; font-family: 'Segoe UI', Roboto, sans-serif; color: #111827; border-bottom: 2px solid #e5e7eb; padding-bottom: 6px;">📘 2. CSV – Comma-Separated Values</h2>
<p style="font-family: 'Segoe UI', Roboto, sans-serif; color: #1f2937; line-height: 1.65; margin: 8px 0 12px;">CSV je jednoduchý <strong>textový formát</strong>, v ktorom jeden riadok typicky predstavuje jeden záznam a jednotlivé hodnoty sú oddelené oddeľovačom.</p>
<p style="font-family: 'Segoe UI', Roboto, sans-serif; color: #1f2937; line-height: 1.65; margin: 8px 0 12px;">Názov síce hovorí <em>Comma-Separated Values</em>, ale v praxi sa môžete stretnúť aj s inými oddeľovačmi, napríklad:</p>
<ul style="font-family: 'Segoe UI', Roboto, sans-serif; color: #1f2937; line-height: 1.7; margin-top: 6px;">
<li>čiarka <span style="background-color: #f2f2f2; font-family: monospace; padding: 2px 4px; border-radius: 3px;">,</span>,</li>
<li>bodkočiarka <span style="background-color: #f2f2f2; font-family: monospace; padding: 2px 4px; border-radius: 3px;">;</span>,</li>
<li>tabulátor.</li>
</ul>
<p style="font-family: 'Segoe UI', Roboto, sans-serif; color: #1f2937; line-height: 1.65; margin: 8px 0 12px;">Pri načítaní cudzieho CSV preto nie je dôležité iba vedieť, že ide o CSV. Potrebujeme poznať aj jeho <strong>delimiter, encoding, hlavičku a spôsob reprezentácie chýbajúcich hodnôt</strong>.</p>

<h2 style="margin: 24px 0 16px; font-size: 20px; font-weight: 600; font-family: 'Segoe UI', Roboto, sans-serif; color: #111827; border-bottom: 2px solid #e5e7eb; padding-bottom: 6px;">📘 3. Apache Parquet</h2>
<p style="font-family: 'Segoe UI', Roboto, sans-serif; color: #1f2937; line-height: 1.65; margin: 8px 0 12px;">Parquet je <strong>binárny stĺpcový formát</strong> navrhnutý pre efektívne ukladanie a analytické čítanie veľkých datasetov.</p>
<p style="font-family: 'Segoe UI', Roboto, sans-serif; color: #1f2937; line-height: 1.65; margin: 8px 0 12px;">CSV ukladá dáta po riadkoch. Parquet ich organizuje po stĺpcoch. To je výhodné najmä pri analytických úlohách, kde často potrebujeme pracovať iba s niekoľkými stĺpcami z veľmi širokej tabuľky.</p>
<p style="font-family: 'Segoe UI', Roboto, sans-serif; color: #1f2937; line-height: 1.65; margin: 8px 0 12px;">Parquet navyše uchováva:</p>
<ul style="font-family: 'Segoe UI', Roboto, sans-serif; color: #1f2937; line-height: 1.7; margin-top: 6px;">
<li>dátové typy,</li>
<li>schému,</li>
<li>metadata,</li>
<li>štatistiky jednotlivých blokov,</li>
<li>kompresiu.</li>
</ul>
<p style="font-family: 'Segoe UI', Roboto, sans-serif; color: #1f2937; line-height: 1.65; margin: 8px 0 12px;">Preto je veľmi bežný v prostrediach ako Spark, Hadoop, dátové jazerá a cloudové analytické platformy.</p>

<p style="font-family: 'Segoe UI', Roboto, sans-serif; color: #1f2937; line-height: 1.65; margin: 8px 0 12px;">Predstavte si dataset so 100 miliónmi riadkov a 80 stĺpcami.</p>
<p style="font-family: 'Segoe UI', Roboto, sans-serif; color: #1f2937; line-height: 1.65; margin: 8px 0 12px;">Ak potrebujeme iba stĺpce <span style="background-color: #f2f2f2; font-family: monospace; padding: 2px 4px; border-radius: 3px;">customer_id</span> a <span style="background-color: #f2f2f2; font-family: monospace; padding: 2px 4px; border-radius: 3px;">sales</span>, pri CSV typicky čítame celý riadkový súbor. Pri Parquete môže engine načítať iba potrebné stĺpce.</p>
<p style="font-family: 'Segoe UI', Roboto, sans-serif; color: #1f2937; line-height: 1.65; margin: 8px 0 12px;">Tento princíp sa nazýva <strong>column pruning</strong> a je jedným z dôvodov, prečo je Parquet vhodný pre analytické workloady.</p>

<h3 style="margin: 20px 0 10px; font-size: 17px; font-weight: 600; font-family: 'Segoe UI', Roboto, sans-serif; color: #111827;">3.2 Konverzia CSV do Parquet</h3>
<p style="font-family: 'Segoe UI', Roboto, sans-serif; color: #1f2937; line-height: 1.65; margin: 8px 0 12px;">Tento príklad slúži na ukážku rozdielu medzi formátmi. CSV načítame do DataFrame a následne ho zapíšeme vo formáte Parquet.</p>
<p style="font-family: 'Segoe UI', Roboto, sans-serif; color: #1f2937; line-height: 1.65; margin: 8px 0 12px;">V reálnej pipeline však môžeme dostať Parquet <strong>priamo zo zdrojového systému</strong> a žiadna konverzia z CSV nemusí prebehnúť.</p>


```python
import pandas as pd

df = pd.read_csv(r"sales_report+(1).csv")

df.to_parquet(
    "sales_report.parquet",
    engine="pyarrow",
    index=False
)

print("Sales dataset saved as Parquet")
```

<h3 style="margin: 20px 0 10px; font-size: 17px; font-weight: 600; font-family: 'Segoe UI', Roboto, sans-serif; color: #111827;">3.3 Načítanie Parquet súboru</h3>
<p style="font-family: 'Segoe UI', Roboto, sans-serif; color: #1f2937; line-height: 1.65; margin: 8px 0 12px;">Na úrovni pandas je čítanie jednoduché <span style="background-color: #f2f2f2; font-family: monospace; padding: 2px 4px; border-radius: 3px;">read_parquet()</span> vráti DataFrame.</p>
<p style="font-family: 'Segoe UI', Roboto, sans-serif; color: #1f2937; line-height: 1.65; margin: 8px 0 12px;">Parquet však obsahuje viac informácií než iba samotné hodnoty. K schéme a metadátam sa dostaneme neskôr pomocou <span style="background-color: #f2f2f2; font-family: monospace; padding: 2px 4px; border-radius: 3px;">pyarrow</span>.</p>


```python
data = pd.read_parquet("sales_report.parquet", engine="pyarrow")

print("Dataset loaded from Parquet")
print(data.head(30))
```

<h3 style="margin: 20px 0 10px; font-size: 17px; font-weight: 600; font-family: 'Segoe UI', Roboto, sans-serif; color: #111827;">3.4 Načítanie iba vybraných stĺpcov</h3>
<p style="font-family: 'Segoe UI', Roboto, sans-serif; color: #1f2937; line-height: 1.65; margin: 8px 0 12px;">Pri Parquete nemusíme načítať všetky stĺpce. Parameter <span style="background-color: #f2f2f2; font-family: monospace; padding: 2px 4px; border-radius: 3px;">columns</span> umožní vybrať iba tie, ktoré potrebujeme.</p>
<p style="font-family: 'Segoe UI', Roboto, sans-serif; color: #1f2937; line-height: 1.65; margin: 8px 0 12px;">Toto je dôležité pri veľkých datasetoch, znižujeme množstvo dát, ktoré musí engine prečítať.</p>


```python
df_small = pd.read_parquet(
    "sales_report.parquet",
    columns=["OrderID", "SalesAmount"]
)

print(df_small.head(10))
```

<h3 style="margin: 20px 0 10px; font-size: 17px; font-weight: 600; font-family: 'Segoe UI', Roboto, sans-serif; color: #111827;">3.5 Vytvorenie nového Parquet datasetu</h3>
<p style="font-family: 'Segoe UI', Roboto, sans-serif; color: #1f2937; line-height: 1.65; margin: 8px 0 12px;">Nasledujúci blok vytvorí jednoduchý dataset a uloží ho do Parquet súboru. Cieľom je ukázať zápis dát a následné opätovné čítanie.</p>


```python
import pandas as pd
import numpy as np

data = {
    "OrderId": range(1, 101),
    "Product": np.random.choice(["Laptop", "Phone", "Tablet", "Monitor"], 100),
    "Category": np.random.choice(["Electronics", "Accessories"], 100),
    "Quantity": np.random.randint(1, 10, 100),
    "Price": np.random.randint(200, 2000, 100),
    "Region": np.random.choice(["North", "South", "East", "West"], 100),
    "Salesperson": np.random.choice(["Alice", "Bob", "Noah", "Diana"], 100)
}

df_new = pd.DataFrame(data)
df_new["SalesAmount"] = df_new["Quantity"] * df_new["Price"]

df_new.to_parquet("sales_report2.parquet", engine="pyarrow", index=False)

print("New Parquet dataset created.")
```

```python
df_check = pd.read_parquet("sales_report2.parquet")
print(df_check.tail(10))
```

<h3 style="margin: 20px 0 10px; font-size: 17px; font-weight: 600; font-family: 'Segoe UI', Roboto, sans-serif; color: #111827;">3.6 Pridanie ďalších dát</h3>
<p style="font-family: 'Segoe UI', Roboto, sans-serif; color: #1f2937; line-height: 1.65; margin: 8px 0 12px;">Samotný Parquet súbor nie je databázová tabuľka, do ktorej jednoducho vykonáme <span style="background-color: #f2f2f2; font-family: monospace; padding: 2px 4px; border-radius: 3px;">INSERT</span>.</p>
<p style="font-family: 'Segoe UI', Roboto, sans-serif; color: #1f2937; line-height: 1.65; margin: 8px 0 12px;">Jedna z možností je:</p>
<ol style="font-family: 'Segoe UI', Roboto, sans-serif; color: #1f2937; line-height: 1.7; margin-top: 6px;">
<li>existujúci súbor načítať,</li>
<li>vytvoriť nové riadky,</li>
<li>spojiť ich pomocou <span style="background-color: #f2f2f2; font-family: monospace; padding: 2px 4px; border-radius: 3px;">concat()</span>,</li>
<li>výsledok znovu zapísať.</li>
</ol>
<p style="font-family: 'Segoe UI', Roboto, sans-serif; color: #1f2937; line-height: 1.65; margin: 8px 0 12px;">Pri veľkých dátových jazerách sa však zvyčajne pracuje s viacerými Parquet súbormi alebo s nadstavbovými tabuľkovými formátmi.</p>

```python
import pandas as pd
import numpy as np

df_old = pd.read_parquet("sales_report2.parquet")

new_data = {
    "OrderId": range(df_old["OrderId"].max() + 1, df_old["OrderId"].max() + 11),
    "Product": np.random.choice(["Laptop", "Phone", "Tablet", "Monitor"], 10),
    "Category": np.random.choice(["Electronics", "Accessories"], 10),
    "Quantity": np.random.randint(1, 10, 10),
    "Price": np.random.randint(200, 2000, 10),
    "Region": np.random.choice(["North", "South", "East", "West"], 10),
    "Salesperson": np.random.choice(["Alice", "Bob", "Noah", "Diana"], 10)
}

df_append = pd.DataFrame(new_data)
df_append["SalesAmount"] = df_append["Quantity"] * df_append["Price"]

df_combined = pd.concat([df_old, df_append], ignore_index=True)

df_combined.to_parquet("sales_report2.parquet", engine="pyarrow", index=False)

print("New rows added and Parquet file rewritten.")
```

<h3 style="margin: 20px 0 10px; font-size: 17px; font-weight: 600; font-family: 'Segoe UI', Roboto, sans-serif; color: #111827;">3.8 Partitioning</h3>
<p style="font-family: 'Segoe UI', Roboto, sans-serif; color: #1f2937; line-height: 1.65; margin: 8px 0 12px;">Pri veľkých datasetoch môžeme dáta rozdeliť do fyzických adresárov podľa hodnoty vybraného stĺpca.</p>
<p style="font-family: 'Segoe UI', Roboto, sans-serif; color: #1f2937; line-height: 1.65; margin: 8px 0 12px;">Ak partitionujeme podľa <span style="background-color: #f2f2f2; font-family: monospace; padding: 2px 4px; border-radius: 3px;">Region</span>, vznikne štruktúra približne:</p>
<pre style="background: #f9fafb; padding: 16px; border-radius: 8px; overflow: auto; font-size: 14px; line-height: 1.55; font-family: ui-monospace, SFMono-Regular, Menlo, monospace; margin: 12px 0 16px;">partitioned_sales/
├── Region=East/
├── Region=North/
├── Region=South/
└── Region=West/
</pre>
<p style="font-family: 'Segoe UI', Roboto, sans-serif; color: #1f2937; line-height: 1.65; margin: 8px 0 12px;">Ak neskôr potrebujeme iba <span style="background-color: #f2f2f2; font-family: monospace; padding: 2px 4px; border-radius: 3px;">Region = West</span>, analytický engine môže preskočiť ostatné priečinky. Tento princíp sa nazýva <strong>partition pruning</strong>.</p>

```python
import pandas as pd
import os

df = pd.read_csv(r"sales_report+(1).csv")

df["Date"] = pd.to_datetime(df["Date"])

output_dir = r"partitioned_sales"
os.makedirs(output_dir, exist_ok=True)

df.to_parquet(
    output_dir,
    engine="pyarrow",
    partition_cols=["Region"],
    index=False,
    coerce_timestamps="us",
    allow_truncated_timestamps=True
)

print("Partitioned Parquet dataset saved.")
```

<p style="font-family: 'Segoe UI', Roboto, sans-serif; color: #1f2937; line-height: 1.65; margin: 8px 0 12px;">Každý row group obsahuje časť jázd a vo vnútri sú jeho stĺpce uložené samostatne.</p>

<p style="font-family: 'Segoe UI', Roboto, sans-serif; color: #1f2937; line-height: 1.65; margin: 8px 0 12px;">Pozrime sa na prvý row group.</p>
<p style="font-family: 'Segoe UI', Roboto, sans-serif; color: #1f2937; line-height: 1.65; margin: 8px 0 12px;">Nečítame ešte konkrétne jazdy. Zisťujeme iba technické informácie:</p>
<ul style="font-family: 'Segoe UI', Roboto, sans-serif; color: #1f2937; line-height: 1.7; margin-top: 6px;">
<li>koľko riadkov obsahuje,</li>
<li>koľko má stĺpcov,</li>
<li>akú veľkosť predstavuje.</li>
</ul>
<p style="font-family: 'Segoe UI', Roboto, sans-serif; color: #1f2937; line-height: 1.65; margin: 8px 0 12px;">Potom rovnaké údaje vypíšeme pre všetky row groups a uvidíme, ako je januárový dataset rozdelený.</p>

```python
row_group = metadata.row_group(0)

print("Rows:", row_group.num_rows)
print("Columns:", row_group.num_columns)
print("Total byte size:", row_group.total_byte_size)
```

<p style="font-family: 'Segoe UI', Roboto, sans-serif; color: #1f2937; line-height: 1.65; margin: 8px 0 12px;">Teraz si môžeme porovnať všetky row groups v súbore.</p>
<p style="font-family: 'Segoe UI', Roboto, sans-serif; color: #1f2937; line-height: 1.65; margin: 8px 0 12px;">Takýto prehľad nám ukáže, ako je dataset fyzicky rozdelený.</p>

```python
for i in range(metadata.num_row_groups):
    rg = metadata.row_group(i)
    print(
        f"Row group {i}: "
        f"rows={rg.num_rows:,}, "
        f"columns={rg.num_columns}, "
        f"size={rg.total_byte_size / 1024 / 1024:.2f} MB"
    )
```

<p style="font-family: 'Segoe UI', Roboto, sans-serif; color: #1f2937; line-height: 1.65; margin: 8px 0 12px;">V rámci jedného row group môžeme ísť ešte hlbšie a pozrieť sa na konkrétny stĺpec.</p>
<p style="font-family: 'Segoe UI', Roboto, sans-serif; color: #1f2937; line-height: 1.65; margin: 8px 0 12px;">Pri column chunk nás nezaujímajú jednotlivé hodnoty jázd, ale napríklad:</p>
<ul style="font-family: 'Segoe UI', Roboto, sans-serif; color: #1f2937; line-height: 1.7; margin-top: 6px;">

<li>použitá kompresia,</li>
<li>počet uložených hodnôt,</li>
<li>komprimovaná a nekomprimovaná veľkosť.</li>
</ul>
<p style="font-family: 'Segoe UI', Roboto, sans-serif; color: #1f2937; line-height: 1.65; margin: 8px 0 12px;">Takto vieme vidieť, ako Parquet fyzicky ukladá jednotlivé stĺpce.</p>

```python
column = metadata.row_group(0).column(0)

print("Column:", column.path_in_schema)
print("Physical type:", column.physical_type)
print("Compression:", column.compression)
print("Number of values:", column.num_values)
print("Compressed size:", column.total_compressed_size)
print("Uncompressed size:", column.total_uncompressed_size)
```

<h2 style="margin: 24px 0 16px; font-size: 20px; font-weight: 600; font-family: 'Segoe UI', Roboto, sans-serif; color: #111827; border-bottom: 2px solid #e5e7eb; padding-bottom: 6px;">📘 4.6 Minimum, maximum a NULL hodnoty bez načítania stĺpca</h2>
<p style="font-family: 'Segoe UI', Roboto, sans-serif; color: #1f2937; line-height: 1.65; margin: 8px 0 12px;">Pri niektorých stĺpcoch Parquet ukladá aj štatistiky.</p>
<p style="font-family: 'Segoe UI', Roboto, sans-serif; color: #1f2937; line-height: 1.65; margin: 8px 0 12px;">Na stĺpci <span style="background-color: #f2f2f2; font-family: monospace; padding: 2px 4px; border-radius: 3px;">trip_distance</span> si vieme pozrieť napríklad:</p>
<ul style="font-family: 'Segoe UI', Roboto, sans-serif; color: #1f2937; line-height: 1.7; margin-top: 6px;">
<li>minimálnu vzdialenosť,</li>
<li>maximálnu vzdialenosť,</li>
<li>počet <span style="background-color: #f2f2f2; font-family: monospace; padding: 2px 4px; border-radius: 3px;">NULL</span> hodnôt.</li>
</ul>
<p style="font-family: 'Segoe UI', Roboto, sans-serif; color: #1f2937; line-height: 1.65; margin: 8px 0 12px;">Dôležité je, že tieto hodnoty čítame z metadát. Nepočítame ich teraz z celého stĺpca pomocou pandas.</p>


```python
stats = column.statistics

print(stats)
```

<p style="font-family: 'Segoe UI', Roboto, sans-serif; color: #1f2937; line-height: 1.65; margin: 8px 0 12px;">Pri číselnom stĺpci sú minimum a maximum zaujímavejšie.
```python
column_names = parquet_file.schema_arrow.names
trip_distance_index = column_names.index("trip_distance")

trip_column = metadata.row_group(0).column(trip_distance_index)
stats = trip_column.statistics

print("Column:", trip_column.path_in_schema)
print("Compression:", trip_column.compression)

if stats is not None:
    print("Min:", stats.min)
    print("Max:", stats.max)
    print("Null count:", stats.null_count)
else:
    print("Statistics are not available.")
```

<li>načítajte iba stĺpce, ktoré na jej zodpovedanie potrebujete,</li>
<li>vykonajte potrebnú transformáciu alebo agregáciu,</li>

<div style="margin-top: 12px; background-color: #f9fafb; border-left: 4px solid #60a5fa; padding: 12px; border-radius: 4px; font-family: 'Segoe UI', Roboto, sans-serif; color: #1f2937; line-height: 1.6;">💡 <strong>Zamyslite sa:</strong> Keby bol tento súbor súčasťou ETL pipeline, ktoré z krokov, ktoré ste práve vykonali, by ste chceli automatizovať?</div>

```python
# Vaše riešenie

#Aký je preferovaný spôsob platby zákazníkov a ako sa to prejavuje na priemernej vzdialenosti a sume?
import pandas as pd

stlpce = ["payment_type", "trip_distance", "total_amount"]
df_analysis = pd.read_parquet("green_tripdata_2024-01.parquet", columns=stlpce)

result = df_analysis.groupby("payment_type").agg(
    priemerna_vzdialenost=("trip_distance", "mean"),
    priemerna_suma=("total_amount", "mean"),
    pocet_jazd=("total_amount", "count")
).reset_index()

result["podiel_jazd"] = result["pocet_jazd"] / result["pocet_jazd"].sum() * 100

display(result)

print("Z výsledku vyplýva, že typ platby 1.0 je využívaný najviac a jeho priemerná suma a vzdialenosť je najväčšia.")
```

<p style="font-family: 'Segoe UI', Roboto, sans-serif; color: #1f2937; line-height: 1.65; margin: 8px 0 12px;">Po tejto časti by ste nemali Parquet vnímať iba ako súbor, ktorý otvoríme cez <span style="background-color: #f2f2f2; font-family: monospace; padding: 2px 4px; border-radius: 3px;">pd.read_parquet()</span>.</p>
<p style="font-family: 'Segoe UI', Roboto, sans-serif; color: #1f2937; line-height: 1.65; margin: 8px 0 12px;">Mali by ste vedieť:</p>
<ul style="font-family: 'Segoe UI', Roboto, sans-serif; color: #1f2937; line-height: 1.7; margin-top: 6px;">
<li>pozrieť si jeho schému ešte pred načítaním dát,</li>
<li>zistiť počet riadkov a row groups,</li>
<li>pristúpiť k metadátam konkrétneho row group,</li>
<li>pozrieť si metadata konkrétneho stĺpca,</li>
<li>zistiť minimum, maximum a počet <span style="background-color: #f2f2f2; font-family: monospace; padding: 2px 4px; border-radius: 3px;">NULL</span> hodnôt, ak sú štatistiky dostupné,</li>
<li>načítať iba jeden row group,</li>
<li>načítať iba vybrané stĺpce.</li>
</ul>
<p style="font-family: 'Segoe UI', Roboto, sans-serif; color: #1f2937; line-height: 1.65; margin: 8px 0 12px;">Pri väčšom Parquet súbore je často zbytočné začínať načítaním všetkých stĺpcov a všetkých riadkov. Najskôr si treba zistiť, čo súbor obsahuje a čo z neho pre danú úlohu skutočne potrebujeme.</p>
